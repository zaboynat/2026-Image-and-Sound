**1. Warm up**

There are various news stories about animal rescues. [Here's one.](https://www.london-fire.gov.uk/incidents/2025/october/dog-rescue-hackney/). Can we find the rescue in our data?

Reading the linked news story, the brigade arrived in Hackney at 2:55pm some day before October 26, 2025 (the date of the publication) to rescue a dog. So we have several logical expressions we can apply to each row of our `lfb` table.

1. Is the rescue in October?
2. Is the year 2025?
3. Is the rescue a dog?
4. Is the Borough Hackney?

```python
# 1. Is the rescue in October?

# make a variable "october" that is True/False if rescue month is 10 or not
# here we use the ".dt" attribute to access the date data in 
# "DateTimeOfCall" -- specifically asking for the month of each rescue,
# keeping only those in october

october = lfb["DateTimeOfCall"].dt.month == 10

# how many octobers do we have?
october.sum()
```

```python
# 2. Is the year 2025?

# make a variable "year2025" that is True/False if rescue year is 2025 or not
# here we use the ".dt" attribute to access the date data in 
# "DateTimeOfCall" -- specifically asking for the year of each rescue,
# keeping only those in 2025

year2025 = lfb["DateTimeOfCall"].dt.year == 2025

# how many octobers do we have?
year2025.sum()
```

```python
# 3. Is the rescue a dog?

# make a variable "dog" that is True/False if rescue is of a dog or not
# this was covered in your last homework

dog = lfb["AnimalGroupParent"].str.lower() == "dog"

# how many dogs do we have?
dog.sum()
```

```python
# 4. Is the Borough Hackney?

# make a variable "hackney" that is True/False if rescue is in Hackney or not
# again, this was in your last homework

hackney = lfb["Borough"].str.lower() == "hackney"

# how many dogs do we have?
hackney.sum()
```

Now combine the four True/False variables with `&` ("and"). Only rows where `october`, `year2025`, `dog` and `hackney` are all `True` are kept. (Remember, if we wanted "or" instead of "and" we would have used `|` the vertical bar.)

```python
# keep only the rows where all four conditions are True
# this is just subsetting with True/False like we saw in the last homework

lfb[october & year2025 & dog & hackney]
```

We see two rescues, but only the second one matches the other details in the news report (Victoria Park Road, arrival at 2:55, etc).

That was a lesson in subsetting! What about [this story](https://www.theguardian.com/world/2025/nov/02/firefighters-charities-animal-rescue-callouts-england) that includes overall statistics about domestic animals over time? 

![telegraph.png](images/telegraph.png)


The Telegraph story quotes a number of statistics. For example, from January 1 to October 31" 2024 there were 716 cats rescued, "which is already more than in any other previous year."

We can check that using subsetting and then calculating some simple statistics. As with the Hackney rescue, we break the question into True/False variables, one per condition.

1. Is the year 2024?
2. Is the month January through October?
3. Is the rescue a cat?

```python
# 1. Is the year 2024?

# make a variable "year" that is True/False if rescue year is 2024 or not
# 
# use .dt to access the datetime data in the DateTimeOfCall column,
# extracting just the year.

year = lfb["DateTimeOfCall"].dt.year == 2024

# how many rescues in 2024?
year.sum()
```

```python
# 2. Is the month January through October?

# make a variable "month" that is True/False if the rescue month is 10 or less
#
# use .dt to access the datetime data in the DateTimeOfCall column,
# extracting just the month.

month = lfb["DateTimeOfCall"].dt.month <= 10

# how many rescues are in January through October (of any year)?
month.sum()
```

```python
# 3. Is the rescue a cat?

# make a variable "cat" that is True/False if rescue is of a cat or not
# (we use lowercase to make "CAT" and "Cat" and "cat" all match.

cat = lfb["AnimalGroupParent"].str.lower() == "cat"

# how many cats do we have?
cat.sum()
```

Now combine the three variables with `&` and add up the cost column for just those rows.

```python
# create a variable "interest" that is True/False: True if all the
# conditions (month, year, animal) are satisfied, and False if at least
# one is off (wrong year or wrong animal, say)
#
# ("interest" is like a new column for "lfb", where each entry corresponds
# to a different rescue

interest = year & month & cat

# sum these up (True -> 1 and False -> 0) to see how many cats we have
# in the time period

interest.sum()
```

As reported, 716! Now we can tally the years of all the rescues in `lfb` using `value_counts()` to compare the total number of cats per year prior to 2024.

```python
# keep just the cats from "lfb"
cats = lfb[cat]

# now use .dt to access the datetime data in the DateTimeOfCall column,
# extracting just the year. We finish applying value_counts()

cats["DateTimeOfCall"].dt.year.value_counts()
```

Notice each year (2009 to 2023) has fewer than 716 cats. Telegraph was right again! If you like, open a window above this one and try to verify some of their other claims!

We didn't finish this exercise in class. Let's make not one map, but 4. Where each map is a count of different animal rescues -- cats, dogs, horses and birds. We'll *facet* the plot by animal type. Here's what I asked the AI.

The AI then created a series of text cells (in yellow) and code cells. Just execute the code cells - you don't have to consult an AI again.

<div style="background-color:#fff3cd; padding:12px; border-radius:4px">

This cell counts the rescues in each borough for each of the four animals with `value_counts()`, as we did for cats. It loops over the animals, keeps just the rows for one animal at a time, and adds the counts to `boroughs` as four new columns: `Cat`, `Dog`, `Horse` and `Bird`. As before, the borough names are put in capitals just for the matching, and a borough with no rescues of that animal gets 0.

</div>

```python
# "boroughs" already holds the outlines and the CatRescues column

# the animal type for each row of lfb, in lower case with extra spaces removed
animal_lower = lfb["AnimalGroupParent"].str.strip().str.lower()

for animal in ["cat", "dog", "horse", "bird"]:
    # keep just the rows for this animal
    animal_rows = lfb[animal_lower == animal]

    # number of rescues for each borough name, with the names in capitals
    counts = animal_rows["Borough"].str.upper().value_counts()

    # look up each borough's count by its (capitalized) name; use 0 if there is none
    boroughs[animal.capitalize()] = boroughs["name"].str.upper().map(counts).fillna(0).astype(int)

boroughs[["name", "Cat", "Dog", "Horse", "Bird"]]
```

<div style="background-color:#fff3cd; padding:12px; border-radius:4px">

This cell draws four choropleth maps in a 2 × 2 grid, one for each animal. (The code makes one map per animal in a loop and arranges them in two columns.) **Each map has its own color scale**, because horse rescues would look almost blank next to cats if they shared one. So you can compare *where* each animal is rescued, but a color in one map does not mean the same count as that color in another. Check each legend, or hover over a borough to see its count.

</div>

```python
# convert to the British National Grid (meters) and simplify the outlines a little
# (to within 100 meters) so the four maps stay small
four_bng = boroughs.to_crs(27700)
four_bng["geometry"] = four_bng["geometry"].simplify(100)

maps = []

for animal in ["Cat", "Dog", "Horse", "Bird"]:
    one_map = alt.Chart(four_bng).mark_geoshape(stroke="black", strokeWidth=0.5).encode(
        color=alt.Color(
            animal + ":Q",
            scale=alt.Scale(scheme="yelloworangered"),
            title="Rescues",
        ),
        tooltip=[
            alt.Tooltip("name:N", title="Borough"),
            alt.Tooltip(animal + ":Q", title=animal + " rescues"),
        ],
    ).project(type="identity", reflectY=True).properties(title=animal, width=300, height=250)

    maps.append(one_map)

# 2 columns, so the four maps make 2 rows; each map keeps its own color scale
alt.concat(*maps, columns=2).resolve_scale(color="independent").properties(
    title="Rescues by London borough, by animal, London Fire Brigade"
)
```

So each animal has a different pattern. Horses, for example, are primarily in what must be more suburban/urban areas. These kinds of "small multiples" or facets can help explain a situation.

This cell makes a histogram of how far apart the two daters are in age. It subtracts `age2` from `age1` and takes the absolute value, so a 28-year-old and a 24-year-old are 4 years apart whichever way round it is. Altair then groups the differences into bins 1 wide: [0, 1), [1, 2), [2, 3) and so on. A square bracket means the edge is included and a round bracket means it is not, so a bin holds its left edge but not its right edge. Ages are whole numbers, so each bin holds just one difference (the [3, 4) bar is the dates that are exactly 3 years apart). The bins run from 0 to 14 because the biggest age gap in the data is 13. Every date has both ages, so none are left out; the cell prints how many are missing to show that.

</div>

```python
import altair as alt

# calculate the absolute difference between the two ages, one value per date
age_gap = (dates["age1"] - dates["age2"]).abs()

# every date should have both ages, so this should be 0
print("dates with a blank age:", age_gap.isna().sum())

# altair wants a data frame, so put the gaps in a dictionary then make the dataframe
age_gap_df = pd.DataFrame({"age_gap": age_gap})

# histogram with bins 1 wide, from 0 to 14 (the biggest gap is 13)
alt.Chart(age_gap_df).mark_bar().encode(
    alt.X("age_gap:Q", bin=alt.Bin(step=1, extent=[0, 14]), title="absolute difference between the two ages (years)"),
    alt.Y("count()", title="number of dates"),
).properties(title="How far apart in age are blind daters?", width=500, height=300)
```

**Compiling text**

The data set `small_dates.csv` was built using the Guardian's API (Application Programming Interface). An API is a structured way for computers to exchange information. It defines how you ask for information and the format that information is returned in. The lingua franca of APIs is JSON.

where the parameters we specify are `show-fields` to specify the data about each date we get back. Then the `tag` is set to the blind date series in the life and style section of the paper. And finally there's an API key. The key (like your OpenRouter key) identifies you as you... and, for example, is used to set how much you can download in what period of time. The code below grabs the URL we just examined and brings it into Python.

```python
from requests import get
import pandas as pd

# notice the two ways we imported tools from packages

url = "https://content.guardianapis.com/search?show-fields=all&tag=lifeandstyle/series/blind-date&api-key=0797ee10-34cc-4d0a-9da1-73ecccab2edb"
api = get(url)

print(api.status_code)

result = api.json()
result
```

**Patterns in text**

Finally, we can write some code to pull data from the Guardian's text. This cell finds the names of the two people on each date. The `trailText` column looks like "Lucy, 40, a general manager, meets Leon, 45, ...", so the name before "meets" is the first person and the name after it is the second person. The names go in new columns `name1` and `name2` of `fields_df`.

While there are several ways to define patterns in text, this simple example of a commmon phrase being repeated at the start of each date... this is the kind of thing we can look for (or the AI can look for) when extracting data. There are several ways to pull the name and age of our daters... this code is one way. Asking the LLM directly is another. Advantages?

So-called web scraping looks for patterns in web pages using the text as well as the pattern of tags. This is old magic, but also a lot of fun.

```python
name1_list = []
name2_list = []

for trail in fields_df["trailText"]:
    # the trail looks like "Lucy, 40, a general manager, meets Leon, 45, an operations manager"
    left, right = trail.split(" meets ")

    # the name is the text before the first comma. One trail says "David (left)",
    # so we also drop anything that starts with a bracket
    name1_list.append(left.split(",")[0].split(" (")[0])
    name2_list.append(right.split(",")[0].split(" (")[0])

fields_df["name1"] = name1_list
fields_df["name2"] = name2_list

fields_df[["name1", "name2"]]
```
