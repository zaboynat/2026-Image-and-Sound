***Dictionaries***

We will start with a *dictionary*. In the code cell below, we collect some data about the first couple commenters on the Census Bureau's proposal. Specifically we highlight [one pro](https://www.regulations.gov/comment/USBC-2026-0628-0003) and [one con](https://www.regulations.gov/comment/USBC-2026-0628-0085). From the web page for a comment, what kind of data might we pull? And how will we store it? **In steps a dictionary!**

```python
# you define a dictionary using curly braces around a set of "key-value" pairs

con = {
               "documentType" : "Public Submission",
               "postedDate" : "Sep 11, 2026",
               "commentID" : "USBC-2026-0628-0085",
               "trackingNumber" : "mtx-9fve-q5px",
               "format" : "PDF",
               "pageCount" : 2,
               "link" : "https://www.regulations.gov/comment/USBC-2026-0628-0085",
               "author" : "HOGAR DEL INMIGRANTE,INC",
      }

# have a look
con
```

The data being stored (values on the right) are our basic data types - an integer, some strings.

In terms how we work with dictionaries in Python, think  for a moment to how you use a literal dictionary (Websters?). Finding a definition means specifying the word we are after. This is the idea behind a dictionary in Python — store data (*values*) according to a name (a word or some kind of *key*). The result is a collection of *key-value pairs*.

The curly braces (not parentheses!) in the cell above mean we are creating a dictionary by defining the  set of key-value pairs. The names we give to the data (the word, if you will, we associate with the dictionary entry) are `'format'`,  `'link'`, `'postedDate'` and `'commentID'` (on the left of the colons), among others, and the values, the data, are on the right. 

If we want to lookup or access data, we provide a name in *square brackets.*

```python
con['author']
```

```python
con['format']
```

And you can combine data through, say, addition.

```python
# or combine three party affiliation totals
# (and compare to the integer in the "total" key)

con["documentType"]+" "+con["trackingNumber"]
# notice the auto-complete suggesting keys where we've stored data
```

As a container object (an object that holds or groups other objects), the most obvious set of operations you would like to perform with a list should involve storing and retrieveing data. As we said, a list stores objects in a well-defined order. There is a first, a second, a third, and so on. You access these objects using **an index.**  A small catch: Python refers to positions in the list starting at 0 and not at 1. So the first object has index 0, the second has index 1 and so on.

Again, with a dictionary, you access data by name (key) and with a list you use numbers (entry order).


```python
title
```

```python
# the first element (remember, we start counting from 0)
title[0]
```

```python
# the third element - remember we count 0, 1, 2
title[2]
```

```python
# the third to the end
title[2:]
```

```python
# from the beginning up to but not including the fourth element
title[:3]
```

```python
# from the second element up to but not including the fourth
title[1:3]
```

***APIs***

An API specifies rules whereby we can request data from a server. Many web sites serve their content from databases, including the one powering regulations.gov. In this case, you see hints of it all over.

You access an API like you might a web page. Although instead of formatting data in HTML, it is formatted in JSON (JavaScript Object Notation) or XML. regulations.gov uses JSON which will look familiar to you.

Here we make the request for the comment from Dan Stevenson. (His comment has an id FWS-R6-ES-2024-0186-0022.)
